```python
import gradio as gr
```

```python
def business_assistant(query):
    query = query.lower()

    if "sales" in query:
        return """
📊 Today's Sales

Total Sales: ₹50,000
Orders: 120
Revenue: ₹4,50,000
"""

    elif "inventory" in query:
        return """
📦 Inventory Status

✅ Product A - 25 units
⚠️ Product B - 5 units (Low Stock)
❌ Product C - Out of Stock
"""

    elif "employee" in query:
        return """
👨‍💼 Employee Details

Present: 48
Absent: 2
"""

    elif "report" in query:
        return """
📈 Business Report

✔ Sales Increased by 15%
✔ Customer Satisfaction: 92%
✔ AI Recommendation:
Restock Product B.
"""

    else:
        return "🤖 Ask me about sales, inventory, employees, or reports."
```

```python
demo = gr.Interface(
    fn=business_assistant,
    inputs=gr.Textbox(label="Ask a Business Question"),
    outputs=gr.Textbox(label="AI Response"),
    title="🤖 AI Business Assistant",
    description="AI Automation for Business"
)

demo.launch()
```
